```python
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon, Rectangle
import numpy as np
import random
```

```python
fig, ax = plt.subplots()

points = [
    (0, 4),
    (1, 4),
    (1, 3),
    (2, 3),
    (2, 2),
    (3, 2),
    (3, 1),
    (4, 1),
    (4, 0),
    (0, 0)
]
polygon = Polygon(points, closed=True, facecolor='lightgrey', edgecolor='black', zorder=1)
ax.add_patch(polygon)

white_square = Rectangle((1, 0), 1, 1, facecolor='white', edgecolor='black', zorder=2)
ax.add_patch(white_square)

ax.set_xlim(0, 4.5)
ax.set_ylim(0, 4.5)
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title('Варіант 7')
ax.grid(True)

plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
def f(x):
    if 0.0 <= x <= 1.0:
        return 1.0
    elif 1.0 < x <= 4.0:
        return 3.0
    else:
        return None
    
def g(x):
    if 1.5 <= x <= 2.5:
        return 2.0
    else:
        return None
```

```python
def bounds(func, lower_bound, upper_bound):
    y = [func(x) for x in np.arange(lower_bound, upper_bound, 0.01)]
    
    return min(y), max(y)
```

```python
def monte_carlo(func, N, x_lower_bound, x_upper_bound):

    y_lower_bound, y_upper_bound = bounds(func, x_lower_bound, x_upper_bound)

    hits  = 0
    for _ in range(N):
        x = random.uniform(x_lower_bound, x_upper_bound)
        y = random.uniform(0, y_upper_bound)

        if f(x) > y:
            hits += 1

    return hits
```

```python
N = 10000
```

```python
f_area = monte_carlo(f, N, 0, 4)
g_area = monte_carlo(g, N, 1.5, 2.5)
```

```python
f_area * 12 / N
```

```python
g_area * 2 / N
```

```python
print(f'Total area: {f_area * 12 / N - g_area * 2 / N}')
```
